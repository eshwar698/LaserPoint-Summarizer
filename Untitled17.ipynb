```python
# Imports
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling
)
from datasets import Dataset
from peft import (
    prepare_model_for_kbit_training,
    LoraConfig,
    get_peft_model
)
from transformers import BitsAndBytesConfig
import torch
import json

# 4-bit quantization setup
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16  # fallback to torch.float16 if needed
)
```

```python
model_path = "/content/drive/MyDrive/Mistral-7B-Instruct-v0.1"

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_path, trust_remote_code=True)
tokenizer.pad_token = tokenizer.eos_token  # Important for padding

# Load model in 4-bit
model = AutoModelForCausalLM.from_pretrained(
    model_path,
    quantization_config=bnb_config,
    device_map="auto",
    trust_remote_code=True
)

# Enable gradient checkpointing + prepare for k-bit (LoRA)
model.gradient_checkpointing_enable()
model = prepare_model_for_kbit_training(model)
```

```python
lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj"],  # works for Mistral
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM"
)

model = get_peft_model(model, lora_config)
```

```python
import pandas as pd
from datasets import Dataset
import json

dataset_path = "/content/drive/MyDrive/datasets/processed_file.jsonl"

# Load JSONL as list of dicts
with open(dataset_path, 'r', encoding='utf-8') as f:
    lines = f.readlines()

records = [json.loads(line) for line in lines]
df = pd.DataFrame(records)  # Convert to DataFrame

# Convert to Hugging Face Dataset
dataset = Dataset.from_pandas(df)

# Preprocessing function for SFT
def format_supervised(example):
    prompt = f"{example['instruction'].strip()}\n\n{example['input'].strip()}"
    response = example['output'].strip()

    tokenized_prompt = tokenizer(prompt, padding="max_length", truncation=True, max_length=512)
    tokenized_response = tokenizer(response, padding="max_length", truncation=True, max_length=128)

    # Important: Mask prompt tokens in labels using -100
    labels = [-100] * len(tokenized_prompt["input_ids"]) + tokenized_response["input_ids"]

    # Final input includes prompt + response
    input_ids = tokenized_prompt["input_ids"] + tokenized_response["input_ids"]
    attention_mask = tokenized_prompt["attention_mask"] + tokenized_response["attention_mask"]

    return {
        "input_ids": input_ids[:640],  # truncate total length
        "attention_mask": attention_mask[:640],
        "labels": labels[:640]
    }

# Apply formatting to the full dataset
tokenized_dataset = dataset.map(format_supervised, remove_columns=dataset.column_names, num_proc=2)
```

```python
#resume training
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="/content/drive/MyDrive/mistral_summarization_lora",
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4,
    num_train_epochs=3,
    learning_rate=2e-4,
    logging_steps=10,
    save_steps=20,
    save_total_limit=2,
    fp16=True,
    bf16=False,
    optim="paged_adamw_8bit",
    label_names=["labels"],
    report_to="none",
    logging_dir="./logs"
)

data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

trainer = Trainer(
    model=model,
    train_dataset=tokenized_dataset,
    args=training_args,
    tokenizer=tokenizer,
    data_collator=data_collator
)

# Resume from latest checkpoint
trainer.train(resume_from_checkpoint=True)
```
